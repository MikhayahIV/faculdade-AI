19. Crie dois arrays de números interios aleatórios de tamanho 20 e calcule o ângulo entre eles(produto escalar normalizado)

```python
a19=np.random.randint(1,10,size=20);
b19=np.random.randint(1,10,size=20);
# dot() serve para calcular o produto escalar
# linalg.norm() calcula o tamanho/comprimento
# com isso pegamos o cosseno do ângulo
cos_ang19=np.dot(a19,b19) / (np.linalg.norm(a19)*np.linalg.norm(b19));
# arccos é o cosseno inverso
# o arccos() é a função inversa do cosseno
# recebe o valor do cosseno e retorna o ângulo correspondete em radianos
# clip() vai garantir que o valor esteja entre -1 e 1
angulo_rad19=np.arccos(np.clip(cos_ang19,-1,1));
# então aqui temos o ângulo em radianos
print("Ângulo(radianos):",angulo_rad19);
# degrees() converte radianos para graus
print("Ângulo(graus):",np.degrees(angulo_rad19));
```

## 2. Exercícios de SciPy

1. Calcule a integral da função $f(x) = x^2\sin(x)$ no intervalo de 0 a $\pi$.

```python
def f1(x):
  return (x**2) * np.sin(x);

resultado1, erro1 = integrate.quad(f1,0,np.pi);
print("Resultado da integral:", resultado1);
print("Erro estimado:", erro1)
```

2. Resolva a equação diferencial $dy/dx = x - y$, com $y(0) = 1$, no intervalo de 0 a 10.

```python
def eq2(x,y):
  return x - y;

solucao2 = integrate.solve_ivp(eq2, [0,10],[1], t_eval=np.linspace(0,10,100))
print("y(10) ≈", solucao2.y[0][-1])
```

3. Encontre as raízes da função $f(x) = x^4 - 3x^3 + 2$ utilizando SciPy.

```python
def f3(x):
  return x**4 - 3*x**3 +2;


chutes = [-2,0,1,3];
raizes3 = sorted(set(round(optimize.fsolve(f3,c)[0],4)for c in chutes));
print("Raízes encontradas:", raizes3)
```

4. Use a função de otimização de SciPy para maximizar $f(x) = -x^2 + 4x + 1$.

```python
def f4(x):
  return -x**2 + 4*x +1;

resultado4 = optimize.minimize(lambda x: -f4(x),0);
print("x que maximiza f:", resultado4.x[0]);
print("Valor máximo de f(x):", -resultado4.fun);
```

7. Use a função de interpolação cúbica de SciPy para interpolar os dados (0,0), (1,2), (2,3), (3,5), (4,4).

```python
x7 = np.array([0,1,2,3,4]);
y7 = np.array([0,2,3,5,4]);
f7 = interpolate.interp1d(x7,y7, kind='cubic');

x7Novos = np.linspace(0,4,50);
y7Novos = f7(x7Novos);

plt.figure();
plt.plot(x7,y7, 'o', label= 'Pontos originais');
plt.plot(x7Novos,y7Novos, '-', label= 'Interpolacao cubica');
plt.title("Interpolacao cubica");
plt.legend();
plt.show();
```

9. Ajuste uma curva quadrática aos pontos (1,2), (2,4), (3,5), (4,7), (5,8).

```python
x9 = np.array([1,2,3,4,5,]);
y9 = np.array([2,4,5,7,8]);
def parabola9(x,a,b,c):
  return a * x**2 + b * x + c;

parametros9, _ = curve_fit(parabola9,x9,y9);
print("Parametros ajustados [a,b,c]:", parametros9);

plt.figure();
plt.plot(x9,y9,'o',label='Dados');
x9Novos = np.linspace(1,5,50);
plt.plot(x9Novos, parabola9(x9Novos, *parametros9), label='ajuste quadratico');
plt.title("ajuste de curva");
plt.legend();
plt.show();
```
