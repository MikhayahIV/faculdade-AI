2. Gere uma matriz 4x4 com números inteiros de 1 a 16 e extraia a última linha

```python
# arange cria uma sequência
# no nosso caso ela começa em 1 e vai até 16(novamente o 17 não conta)
# reshape é para mudar o formato, no nosso caso vai transformar esses 16 números
# em uma matriz com 4 linhas e 4 colunas
mat2 = np.arange(1,17).reshape(4,4);
# o [-1,:] significa "Pegue a última linha de mat2 e todas as suas colunas"
# então ele vai na primeira coluna e pega o 13
# segunda coluna e pega o 14 etc
# 0,: = primeira linha
# :,0 = primeira coluna
# [linha,coluna] + :
ultima_linha = mat2[-1, :];
print("Matriz:\n",mat2);
print("Última linha:",ultima_linha);
```

4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.

```python
arr4 = np.random.randint(0,100, size=100);
# median calcula a mediana dos valores( da arr4 no caso)
mediana4 = np.median(arr4);
# e aqui o stats.mode() (do scipy) é utilizado para descobrir a moda
# keepdims=True é para manter a dimensão do resultado
# Então calcula a moda e mantém o formato do resultado de uma maneira que permite acessar
# ele encontra o mode e count
moda4 = stats.mode(arr4, keepdims=True);
print(arr4);
print("Media:", mediana4);

print("Moda:", moda4.mode[0], "- Frequência:", moda4.count[0])
```

19. Crie dois arrays de números interios aleatórios de tamanho 20 e calcule o ângulo entre eles(produto escalar normalizado)

```python
a19=np.random.randint(1,10,size=20);
b19=np.random.randint(1,10,size=20);
# dot() serve para calcular o produto escalar
# linalg.norm() calcula o tamanho/comprimento
# com isso pegamos o cosseno do ângulo
cos_ang19=np.dot(a19,b19) / (np.linalg.norm(a19)*np.linalg.norm(b19));
# arccos é o cosseno inverso
# o arccos() é a função inversa do cosseno
# recebe o valor do cosseno e retorna o ângulo correspondete em radianos
# clip() vai garantir que o valor esteja entre -1 e 1
angulo_rad19=np.arccos(np.clip(cos_ang19,-1,1));
# então aqui temos o ângulo em radianos
print("Ângulo(radianos):",angulo_rad19);
# degrees() converte radianos para graus
print("Ângulo(graus):",np.degrees(angulo_rad19));
```
